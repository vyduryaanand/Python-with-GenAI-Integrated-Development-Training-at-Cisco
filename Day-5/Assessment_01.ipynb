**PROBLEM STATEMENT**

*Write a script that scans a directory (via sys.argv) for .log files, and for each file uses re to extract and count occurrences of ERROR, WARN, and INFO lines.*

**INPUT**

A directory containing app.log with mixed ERROR/WARN/INFO lines

**EXPECTED OUTPUT**

app.log -> {'ERROR': 2, 'WARN': 1, 'INFO': 5}

**HINTS**

•Use os.listdir() plus a filename filter, or glob for *.log

•One compiled pattern with a named 'level' group can serve every file

•collections.Counter is a clean way to tally the extracted levels

```python
#Create a Sample Log file
import os

os.makedirs("logs", exist_ok=True)

log_data = """INFO Server started
INFO User logged in
ERROR Database failed
WARN High memory usage
INFO Request received
ERROR Connection failed
INFO Request completed
INFO Backup completed
INFO Server stopped
"""

with open("logs/app.log", "w") as file:
    file.write(log_data)

print("Log file created.")
```

```python
#Main Program
import os
import sys
import re
from collections import Counter

# Simulate command-line argument in Colab
sys.argv = ["script.py", "logs"]

directory = sys.argv[1]

pattern = re.compile(r"^(?P<level>ERROR|WARN|INFO)\b")

for filename in os.listdir(directory):

    if filename.endswith(".log"):
        filepath = os.path.join(directory, filename)

        counts = Counter()

        with open(filepath, "r") as file:
            for line in file:
                match = pattern.search(line)

                if match:
                    counts[match.group("level")] += 1

        print(filename, "->", dict(counts))
```

**Explanation**

os.listdir() finds the files in the directory, while sys.argv[1] provides the directory path. The regular expression searches each line for ERROR, WARN, or INFO and captures the level using the named group level. Counter then counts how many times each level occurs.

**Conclusion**

 This exercise combines os for directory handling, sys for command-line arguments, re for pattern matching, and Counter for counting log levels.
