**PROBLEM STATEMENT**

*Write a small script that logs each run's start time to a JSON file (list of ISO timestamps), keeping only the last 5 runs, and prints how many seconds elapsed since the previous run.*

**INPUT**

history.json containing a JSON list of previous ISO timestamps (or missing on first run)

**EXPECTED OUTPUT**

Seconds since last run: 86423.5

History now has 5 entries.

**HINTS**
•Use datetime.now().isoformat() to generate each entry

•Load with a try/except FileNotFoundError for the first-ever run

•Slice the list to the last 5 entries with history[-5:] before saving

```python
import json
from datetime import datetime

filename = "history.json"

# Load previous history
try:
    with open(filename, "r") as file:
        history = json.load(file)
except FileNotFoundError:
    history = []

# Calculate time since previous run
if history:
    previous_time = datetime.fromisoformat(history[-1])
    current_time = datetime.now()

    elapsed = (current_time - previous_time).total_seconds()
    print(f"Seconds since last run: {elapsed:.1f}")
else:
    current_time = datetime.now()
    print("First run - no previous timestamp.")

# Add current timestamp
history.append(current_time.isoformat())

# Keep only the last 5 runs
history = history[-5:]

# Save updated history
with open(filename, "w") as file:
    json.dump(history, file, indent=2)

print(f"History now has {len(history)} entries.")
```

**Explanation**

The program loads the existing timestamps from history.json. If a previous timestamp exists, datetime.fromisoformat() converts it back to a datetime and the difference from the current time is calculated in seconds. The current timestamp is then added to the history.

**Conclusion**

The list is trimmed using history[-5:], so the JSON file always contains at most the latest 5 runs. json provides persistence, while datetime handles timestamps and elapsed-time calculation.
