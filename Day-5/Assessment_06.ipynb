**HANDS-ON ACTIVITY**

Run both versions on a 10,000-item list with time.time() and compare wall-clock time directly.

```python
import time

# 10,000 items
numbers = list(range(10000))

# Version 1: using a for loop
start = time.time()

result1 = []

for number in numbers:
    result1.append(number * 2)

end = time.time()

loop_time = end - start


# Version 2: using list comprehension
start = time.time()

result2 = [number * 2 for number in numbers]

end = time.time()

comprehension_time = end - start


# Compare
print(f"Loop version: {loop_time:.6f} seconds")
print(f"List comprehension version: {comprehension_time:.6f} seconds")
```

**Explanation**

Both versions perform the same calculation on all 10,000 items. time.time() records the start and end time, and their difference gives the wall-clock time taken by each version.

**Conclusion**

The activity demonstrates how to directly measure and compare the execution time of two Python implementations. In many cases, list comprehensions are faster than an equivalent Python for loop for simple list-building operations.
