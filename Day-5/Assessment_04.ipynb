**PROBLEM STATEMENT**

*Build a CLI tool (argparse) that accepts a directory and a --workers count, then uses threading to check (simulated) 'availability' of N files concurrently, printing total elapsed time.*

**INPUT**

python check.py ./data --workers 4

**EXPECTED OUTPUT**

Checked 10 files with 4 workers in 2.5s (vs ~10s sequential)

**HINTS**

•Each simulated check can just time.sleep(1) and print a result

•Split the file list into chunks, or launch one thread per file up to --workers at a time

•Compare a sequential run's time.time() delta against the threaded version

```python
#Create Sample Files
import os

os.makedirs("data", exist_ok=True)

for i in range(10):
    with open(f"data/file{i+1}.txt", "w") as file:
        file.write("sample data")

print("10 files created.")
```

```python
import argparse
import os
import time
from concurrent.futures import ThreadPoolExecutor

def check_file(filename):
    """Simulate checking a file."""
    time.sleep(1)
    return filename

# Simulate CLI arguments in Colab
parser = argparse.ArgumentParser()
parser.add_argument("directory")
parser.add_argument("--workers", type=int, default=4)

args = parser.parse_args(args=["data", "--workers", "4"])

files = os.listdir(args.directory)

start_time = time.time()

with ThreadPoolExecutor(max_workers=args.workers) as executor:
    results = list(executor.map(check_file, files))

elapsed = time.time() - start_time

print(
    f"Checked {len(results)} files with "
    f"{args.workers} workers in {elapsed:.1f}s"
)
```

**Explanation**

ThreadPoolExecutor creates multiple worker threads that can execute the simulated file checks concurrently. Since each check waits for 1 second, 4 workers can process several files at the same time, reducing the total elapsed time compared with checking all 10 files one after another.

**Conclusion**

This exercise demonstrates basic CLI argument handling with argparse and concurrent execution using threads. time.time() allows us to measure and compare the execution time.
