**PROBLEM STATEMENT**

*Write an automation script that: organizes a downloads folder by extension (process automation), then calls a public REST APIand saves the JSON response to a timestamped file (network automation).*

**INPUT**

A downloads/ folder with mixed file types; a public JSON API URL

**EXPECTED OUTPUT**

Organized 12 files into 3 folders.

Saved API response to snapshot_2026-09-13T10-00-00.json

HINTS
•Reuse the extension-sorting pattern from the Basic Example

•Use datetime.now().strftime(...) to build the timestamped filename

•Wrap the API call in try/except requests.RequestException so one failure doesn't crash the whole script

```python
#Create Sample downloads folder
import os

os.makedirs("downloads", exist_ok=True)

files = [
    "report.pdf",
    "resume.pdf",
    "photo.jpg",
    "image.png",
    "notes.txt",
    "data.csv",
    "marks.csv",
    "readme.txt",
    "logo.jpg",
    "document.pdf",
    "test.png",
    "info.txt"
]

for filename in files:
    with open(os.path.join("downloads", filename), "w") as file:
        file.write("sample")

print("12 sample files created.")
```

```python
#Automation Script
import os
import shutil
import json
import requests
from datetime import datetime

downloads_folder = "downloads"

# Organize files by extension
organized_files = 0
folders = set()

for filename in os.listdir(downloads_folder):

    filepath = os.path.join(downloads_folder, filename)

    if os.path.isfile(filepath):
        extension = os.path.splitext(filename)[1].lower().replace(".", "")

        if extension:
            folder = os.path.join(downloads_folder, extension)
            os.makedirs(folder, exist_ok=True)

            shutil.move(filepath, os.path.join(folder, filename))

            organized_files += 1
            folders.add(extension)

print(f"Organized {organized_files} files into {len(folders)} folders.")

# Call public REST API
api_url = "https://jsonplaceholder.typicode.com/todos/1"

try:
    response = requests.get(api_url, timeout=10)
    response.raise_for_status()

    data = response.json()

    timestamp = datetime.now().strftime("%Y-%m-%dT%H-%M-%S")
    output_file = f"snapshot_{timestamp}.json"

    with open(output_file, "w") as file:
        json.dump(data, file, indent=2)

    print(f"Saved API response to {output_file}")

except requests.RequestException as error:
    print("API request failed:", error)
```

**Explanation**

The script first scans downloads, extracts each file's extension, creates the corresponding folder, and moves the file into it. It then sends a GET request to the REST API using requests, converts the response to JSON, and saves it with the current timestamp in the filename.

**Conclusion**

This exercise combines process automation (os, shutil), network automation (requests), JSON handling, and timestamp-based file naming into one practical Python workflow.
