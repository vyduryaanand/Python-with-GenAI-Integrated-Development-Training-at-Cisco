**PROBLEM STATEMENT**

Read products.csv into a pandas DataFrame, compute total inventory value (price * quantity) per category, and also parse a smallcatalog.xml file to cross-check that every XML product id exists in the CSV. *italicized text*

**INPUT**

products.csv (id, category, price, quantity) and catalog.xml (list of <product id=...>)

**EXPECTED OUTPUT**

Electronics: 4820.00

Home: 1390.50

All XML ids found in CSV.

**HINTS**

•Use df.groupby("category").apply(...) or a computed column then groupby().sum()

•Build a set of CSV ids and check XML ids with `not in` for the cross-check

•int()/float() cast XML text fields before doing arithmetic on them

```python
#import pandas
import pandas as pd
import xml.etree.ElementTree as ET
```

```python
#Create Sample Files
csv_data = """id,category,price,quantity
101,Electronics,1200,2
102,Electronics,605,4
103,Home,463.5,3
"""

with open("products.csv", "w") as file:
    file.write(csv_data)

xml_data = """<catalog>
    <product id="101"/>
    <product id="102"/>
    <product id="103"/>
</catalog>
"""

with open("catalog.xml", "w") as file:
    file.write(xml_data)

print("Files created.")
```

```python
#Process CSV and XML
# Read CSV
df = pd.read_csv("products.csv")

# Calculate inventory value
df["inventory_value"] = df["price"] * df["quantity"]

# Total value per category
category_totals = df.groupby("category")["inventory_value"].sum()

for category, total in category_totals.items():
    print(f"{category}: {total:.2f}")

# Get CSV product IDs
csv_ids = set(df["id"].astype(str))

# Read XML
tree = ET.parse("catalog.xml")
root = tree.getroot()

xml_ids = {product.get("id") for product in root.findall("product")}

# Cross-check IDs
missing_ids = xml_ids - csv_ids

if not missing_ids:
    print("All XML ids found in CSV.")
else:
    print("Missing XML ids:", missing_ids)
```

Explanation

The CSV is loaded into a pandas DataFrame, where inventory_value is calculated as price × quantity. groupby() then calculates the total inventory value for each category. The XML file is parsed using ElementTree, and its product IDs are compared with the CSV IDs using sets.

Conclusion

 This exercise demonstrates how pandas simplifies CSV analysis and how Python's standard XML library can be used to validate data between different file formats.
