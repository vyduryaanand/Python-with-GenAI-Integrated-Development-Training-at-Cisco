**PROBLEM STATEMENT**

*Take a naive function that checks if a list has any duplicate values (nested loop). Ask an AI assistant to optimize it, then benchmark the naive vs. optimized version on a 20,000-item list and report the speedup factor.*

**INPUT**

A list of 20,000 integers with some duplicates

**EXPECTED OUTPUT**

Naive: 6.7s, Optimized: 0.004s, Speedup: ~1675x

**HINTS**

•Use time.time() before/after each version, on the exact same input list

•A set-based membership check is the typical O(n) rewrite

•Report the speedup as a ratio, not just 'it was faster'

```python
import time
import random

# Create a 20,000-item list with duplicates
random.seed(42)
numbers = [random.randint(1, 10000) for _ in range(20000)]


# Naive version - nested loop
def has_duplicate_naive(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return True
    return False


# Optimized version - set
def has_duplicate_optimized(items):
    seen = set()

    for item in items:
        if item in seen:
            return True
        seen.add(item)

    return False


# Benchmark naive version
start = time.time()
result_naive = has_duplicate_naive(numbers)
naive_time = time.time() - start


# Benchmark optimized version
start = time.time()
result_optimized = has_duplicate_optimized(numbers)
optimized_time = time.time() - start


# Calculate speedup
speedup = naive_time / optimized_time

print("Duplicate found:", result_naive)
print(f"Naive: {naive_time:.6f}s")
print(f"Optimized: {optimized_time:.6f}s")
print(f"Speedup: {speedup:.2f}x")
```

**Explanation**

The naive function uses nested loops, giving it O(n²) time complexity in the worst case. The optimized version uses a set, where membership checking is typically O(1) on average, giving an overall O(n) approach.

**Conclusion**

The set-based implementation is significantly more efficient for large lists. The speedup factor is calculated by dividing the naive execution time by the optimized execution time, giving a direct numerical comparison.
