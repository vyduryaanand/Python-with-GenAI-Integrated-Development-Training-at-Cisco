**PROBLEM STATEMENT**

*Given a list of word strings, build: (1) a list of uppercase words longer than 3 characters, (2) a dict mapping each word to itslength, (3) a list of only the palindromes using a generator expression.*

**INPUT**

words = ['python', 'level', 'hi', 'radar', 'ok']

**EXPECTED OUTPUT**

['PYTHON', 'LEVEL', 'RADAR']

{'python': 6, 'level': 5, 'hi': 2, 'radar': 5, 'ok': 2}

['level', 'radar']

**HINTS**
•A palindrome equals its own reverse: w == w[::-1]

•Wrap the generator in list(...) only to print it for verification

•len(w) > 3 is the filter condition for part 1

```python
words = ['python', 'level', 'hi', 'radar', 'ok']

# 1. Uppercase words longer than 3 characters
uppercase_words = [
    w.upper() for w in words
    if len(w) > 3
]

# 2. Dictionary of words and their lengths
word_lengths = {
    w: len(w) for w in words
}

# 3. Generator expression for palindromes
palindromes = (
    w for w in words
    if w == w[::-1]
)

print(uppercase_words)
print(word_lengths)
print(list(palindromes))
```

**Explanation**

The list comprehension filters words whose length is greater than 3 and converts them to uppercase. The dictionary comprehension creates a key-value pair for each word and its length. The generator expression checks whether each word is equal to its reverse using w == w[::-1], producing only the palindrome words.

**Conclusion**

This exercise demonstrates how list comprehensions, dictionary comprehensions, and generator expressions can make Python code concise and readable. Comprehensions are useful for creating new collections from existing data, while generator expressions produce values efficiently when they are needed.
