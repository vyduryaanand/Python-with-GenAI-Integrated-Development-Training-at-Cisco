**PROBLEM STATEMENT**

Take the custom exception-handling function you wrote earlier today (ship_order or transfer_funds) and prompt an AI assistant togenerate a PyTest module for it. Review the output, add one edge case the assistant missed, then run the suite.

**INPUT**

Your ship_order(item, qty, stock) function and its InsufficientStockError

**EXPECTED OUTPUT**

A test_ship_order.py file; 'pytest -v' output showing all tests passing, including at least one edge case you added yourself

**HINTS**

•Ask the assistant explicitly for a test where qty equals stock[item] exactly (a boundary case)

•Verify the assistant used pytest.raises for the exception path, not a manual try/except

•Re-run pytest after every edit, don't just read the diff

```python
%%writefile ship_order.py

class InsufficientStockError(Exception):
    pass


def ship_order(item, qty, stock):
    if qty > stock.get(item, 0):
        raise InsufficientStockError(
            f"Insufficient stock for {item}"
        )

    stock[item] -= qty
    return f"Shipped {qty} {item}"
```

```python
%%writefile test_ship_order.py

import pytest
from ship_order import ship_order, InsufficientStockError


def test_ship_order_success():
    stock = {"laptop": 10}

    result = ship_order("laptop", 3, stock)

    assert result == "Shipped 3 laptop"
    assert stock["laptop"] == 7


def test_ship_order_insufficient_stock():
    stock = {"laptop": 5}

    with pytest.raises(InsufficientStockError):
        ship_order("laptop", 6, stock)


def test_ship_order_missing_item():
    stock = {}

    with pytest.raises(InsufficientStockError):
        ship_order("laptop", 1, stock)


def test_ship_order_exact_stock():
    stock = {"laptop": 5}

    result = ship_order("laptop", 5, stock)

    assert result == "Shipped 5 laptop"
    assert stock["laptop"] == 0
```

```python
!pytest -v
```

**Explanation**

The AI-generated PyTest module tests both normal and exception paths of ship_order(). pytest.raises() verifies that InsufficientStockError is raised when the requested quantity exceeds available stock. The manually added test_ship_order_exact_stock() checks an important boundary condition where the requested quantity is exactly equal to the available stock; this should be valid and reduce the stock to zero. Running pytest -v gives a detailed result for every test, making it easy to identify and fix failures.

**Conclusion**

This exercise demonstrates how an AI assistant can accelerate test-case creation while the developer remains responsible for reviewing the generated tests and identifying missing edge cases. The final suite covers successful orders, insufficient stock, missing items, and the exact-stock boundary condition, with all tests passing.
