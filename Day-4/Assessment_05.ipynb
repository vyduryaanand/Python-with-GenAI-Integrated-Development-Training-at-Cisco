**PROBLEM STATEMENT**

Given a function with a hidden off-by-one bug in an index expression, use breakpoint()/pdb (or your IDE's debugger) to step through, inspect len(items), and identify exactly where the index goes out of range; then read and interpret the resulting traceback.

**INPUT**

def last_two(items):

return [items[len(items)-1], items[len(items)]] # bug

**EXPECTED OUTPUT**

IndexError: list index out of range, pinpointed to the second index expression

**HINTS**

•Step through with n and inspect len(items) with p

•The valid index range is 0 to len(items)-1

•Fix by using items[-1] and items[-2] instead

```python
def last_two(items):
    breakpoint()
    return [items[len(items)-1], items[len(items)]]

items = [10, 20, 30]

print(last_two(items))
```

**Explanation**

The function first accesses items[len(items)-1], which becomes items[2] and is valid because 2 is the last index of a three-element list. The second expression uses items[len(items)], which becomes items[3]; since Python list indexes start at 0, the largest valid index is always len(items)-1, so index 3 is outside the list and Python raises IndexError: list index out of range. The traceback tells us the exception type and points to the exact line where the invalid access occurred.

**Conclusion**

The bug is an off-by-one indexing error. The function can be fixed by using negative indexing:

```python
def last_two(items):
    return [items[-1], items[-2]]

items = [10, 20, 30]

print(last_two(items))
```
