**PROBLEM STATEMENT**

*Model a Shape base class with an area() method, then Circle and Rectangle subclasses that override it. Add __str__ so printing an instance shows its name and area rounded to 2 decimals.*

**INPUT**

shapes = [Circle(3), Rectangle(4, 5)]

**EXPECTED OUTPUT**

Circle: area=28.27

Rectangle: area=20.00

**HINTS**

•Use super().__init__() if Shape stores a shared name attribute

•math.pi is available from the standard math module

•Format with f"{area:.2f}" inside __str__

```python
import math


class Shape:
    def __init__(self, name):
        self.name = name

    def area(self):
        raise NotImplementedError


class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def __str__(self):
        return f"{self.name}: area={self.area():.2f}"


class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__("Rectangle")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def __str__(self):
        return f"{self.name}: area={self.area():.2f}"


shapes = [Circle(3), Rectangle(4, 5)]

for shape in shapes:
    print(shape)
```

**Explanation**

The Shape class acts as the base class and stores the common name attribute. Circle and Rectangle inherit from Shape and override the area() method with their respective formulas. The super().__init__() call initializes the common name, while __str__() formats each object's name and calculated area to two decimal places when the object is printed.

**Conclusion**

This exercise demonstrates key Object-Oriented Programming concepts, including classes, inheritance, method overriding, super(), and the __str__() special method. The same area() method can behave differently for different shape objects, demonstrating polymorphism.
