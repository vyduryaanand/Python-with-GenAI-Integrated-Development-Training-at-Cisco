**Problem Statement**

*Create a reusable inventory.py module containing an InventoryItem class with encapsulated _quantity, restock() and sell() methods, a custom InsufficientStockError, and module-level logging. Then import the class into a separate driver script, demonstrate normal and exception behavior, and create and run a PyTest test module containing normal, exception, and boundary-case tests.*

```python
#Create inventory.py
%%writefile inventory.py

import logging

logger = logging.getLogger(__name__)


class InsufficientStockError(Exception):
    pass


class InventoryItem:
    def __init__(self, name, quantity):
        self.name = name
        self._quantity = quantity

    def restock(self, amount):
        self._quantity += amount
        logger.info(
            f"restocked {self.name} by {amount}, "
            f"quantity={self._quantity}"
        )

    def sell(self, amount):
        if amount > self._quantity:
            logger.error(
                f"cannot sell {amount} {self.name}, "
                f"only {self._quantity} in stock"
            )
            raise InsufficientStockError(
                f"cannot sell {amount} {self.name}, "
                f"only {self._quantity} in stock"
            )

        self._quantity -= amount
        logger.info(
            f"sold {amount} {self.name}, "
            f"quantity={self._quantity}"
        )
```

```python
#Create The driver script
%%writefile driver.py

import logging
from inventory import InventoryItem, InsufficientStockError

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s",
    force=True
)

item = InventoryItem("widget", 10)

item.restock(5)
item.sell(12)
item.sell(3)

try:
    item.sell(1)
except InsufficientStockError:
    pass
```

```python
#Create the AI-generated PyTest module
%%writefile test_inventory.py

import pytest

from inventory import InventoryItem, InsufficientStockError


def test_restock():
    item = InventoryItem("widget", 10)

    item.restock(5)

    assert item._quantity == 15


def test_sell():
    item = InventoryItem("widget", 10)

    item.sell(4)

    assert item._quantity == 6


def test_insufficient_stock():
    item = InventoryItem("widget", 10)

    with pytest.raises(InsufficientStockError):
        item.sell(11)


def test_sell_exact_remaining_stock():
    item = InventoryItem("widget", 10)

    item.sell(10)

    assert item._quantity == 0


def test_sell_one_more_than_remaining_stock():
    item = InventoryItem("widget", 10)

    item.sell(10)

    with pytest.raises(InsufficientStockError):
        item.sell(1)
```

**Explanation**

The InventoryItem class stores the item name and quantity, with _quantity treated as a protected attribute by convention. restock() increases the quantity and logs an INFO message, while sell() checks whether enough stock is available; if not, it logs an ERROR and raises InsufficientStockError, otherwise it decreases the quantity and logs the successful sale. The separate driver.py demonstrates importing and using the class without directly modifying _quantity. The PyTest module verifies normal restocking and selling, insufficient-stock handling, and the important boundary case where selling exactly all remaining stock succeeds.
