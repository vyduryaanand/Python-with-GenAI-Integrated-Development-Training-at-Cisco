**PROBLEM STATEMENT**

*Create a module string_tools.py with functions to reverse a string and count vowels. Import it two ways (import and from...import) into a driver script, then use pip to check whether the 'requests' package is installed.*

**INPUT**

text = "Hello Python Training"

**EXPECTED OUTPUT**

Reversed: gniniarT nohtyP olleH

Vowel count: 6

**HINTS**

•Use import string_tools as st for the aliased form

•pip show requests tells you if/where a package is installed

•vowels = 'aeiouAEIOU' is the filter set for part 2





```python
%%writefile string_tools.py

def reverse_string(text):
    return text[::-1]


def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for char in text:
        if char in vowels:
            count += 1

    return count
```

```python
import string_tools as st
from string_tools import count_vowels

text = "Hello Python Training"

reversed_text = st.reverse_string(text)
vowels = count_vowels(text)

print("Reversed:", reversed_text)
print("Vowel count:", vowels)
```

**Explanation**

In the first cell, %%writefile string_tools.py creates a separate Python module file containing reverse_string() and count_vowels(). In the second cell, import string_tools as st imports the module using the alias st, while from string_tools import count_vowels imports the count_vowels() function directly. The reverse function uses slicing [::-1], and the vowel function checks each character against aeiouAEIOU. Finally, !pip show requests checks the installed details of the requests package.
