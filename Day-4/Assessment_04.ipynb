**PROBLEM STATEMENT**

*Write transfer_funds(from_acct, to_acct, amount) that raises a custom InsufficientFundsError when the source account lacks funds, logs an INFO message on success and an ERROR message on failure (no print), and always logs a finally message noting the attempt completed.*

**INPUT**

accounts = {'A': 100, 'B': 50}

transfer_funds('A', 'B', 30)

transfer_funds('B', 'A', 999)

**EXPECTED OUTPUT**

INFO: transferred 30 from A to B

INFO: transfer attempt complete

ERROR: insufficient funds in B for transfer of 999

INFO: transfer attempt complete

**HINTS**

•Raise a custom InsufficientFundsError(Exception) subclass

•Put the 'attempt complete' log line in finally

•Use logger.error(str(e)) inside the except block

```python
import logging
```

```python
logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s",
    force=True
)

logger = logging.getLogger()
```

```python
class InsufficientFundsError(Exception):
    pass
```

```python
accounts = {
    'A': 100,
    'B': 50
}
```

```python
def transfer_funds(from_acct, to_acct, amount):
    try:
        if accounts[from_acct] < amount:
            raise InsufficientFundsError(
                f"insufficient funds in {from_acct} "
                f"for transfer of {amount}"
            )

        accounts[from_acct] -= amount
        accounts[to_acct] += amount

        logger.info(
            f"transferred {amount} from {from_acct} to {to_acct}"
        )

    except InsufficientFundsError as e:
        logger.error(str(e))

    finally:
        logger.info("transfer attempt complete")
```

```python
transfer_funds('A', 'B', 30)
transfer_funds('B', 'A', 999)
```

**Explanation**

The program first configures Python's logging system using logging.basicConfig(). The format="%(levelname)s: %(message)s" setting ensures that only the log level and message are displayed, while force=True is used because Google Colab may already have a logging configuration. The custom InsufficientFundsError is raised when the source account has less money than the requested transfer amount. The try block performs the transaction and logs successful transfers with logger.info(), while the except block catches the custom exception and logs it using logger.error(). The finally block always executes and logs that the transfer attempt is complete.

**Conclusion**

The earlier message ERROR:__main__:... was not a problem with your transfer logic. It happened because Colab's logging configuration included the logger name __main__. Using format="%(levelname)s: %(message)s" together with force=True gives the exact output required by the exercise.
