**PROBLEM STATEMENT**

*Given a list of (priority, job_name) pairs, use heapq to process jobs in priority order, then use collections.Counter on job categories (derived from job names) to report how many jobs fall in each category.*

**INPUT**

jobs = [(3, "build_report"), (1, "deploy_app"), (2, "review_code"), (1, "deploy_hotfix")]

**EXPECTED OUTPUT**

Jobs printed in ascending priority order, followed by a Counter-based tally grouped by the part of the name before the underscore.

**HINTS**

•heapq.heapify() then repeated heappop() drains jobs in priority order

•job_name.split("_")[0] extracts the category prefix

•Counter accepts any iterable of hashable values, including a generator expression

```python
import heapq
from collections import Counter

jobs = [
    (3, "build_report"),
    (1, "deploy_app"),
    (2, "review_code"),
    (1, "deploy_hotfix")
]

heapq.heapify(jobs)

processed_jobs = []

while jobs:
    priority, job_name = heapq.heappop(jobs)
    print(priority, job_name)
    processed_jobs.append(job_name)

categories = Counter(
    job_name.split("_")[0] for job_name in processed_jobs
)

print(categories)
```

**Explanation**

heapq.heapify() converts the list into a min-heap, so heappop() always removes the job with the lowest priority number first. The job names are then used to extract categories such as deploy, build, and review.

**Conclusion**

heapq is useful for priority-based processing, while Counter provides a simple way to count occurrences of categories in the processed data.
