**PROBLEM STATEMENT**

*Write a function that reads a text file of comma-separated 'name,score' lines, skips any malformed lines, computes the average score, and writes a one-line summary report to a new output file —handling a missing input file gracefully.*

**INPUT**

students.txt containing:

Ana,88

Bo,92

corrupted_line

Cy,79

**EXPECTED OUTPUT**

summary.txt containing something like: 'Processed 3 students, average score: 86.33' and a console message noting the skipped malformed line.

**HINTS**

Use os.path.exists() or try/except FileNotFoundError for the missing-file case

Iterate the input file object directly with a for loop

Open the output file in 'w' mode and use write() with the computed summary string

```python
def process_students(input_file, output_file):
    """Process student scores and write an average summary."""

    total_score = 0
    student_count = 0

    try:
        with open(input_file, "r") as file:
            for line in file:
                line = line.strip()
                parts = line.split(",")

                if len(parts) != 2:
                    print("Skipped malformed line:", line)
                    continue

                name, score = parts

                try:
                    score = float(score)
                except ValueError:
                    print("Skipped malformed line:", line)
                    continue

                total_score += score
                student_count += 1

        average = total_score / student_count if student_count else 0

        summary = f"Processed {student_count} students, average score: {average:.2f}"

        with open(output_file, "w") as file:
            file.write(summary)

        print(summary)

    except FileNotFoundError:
        print("Input file not found.")
```

```python
#Create students.txt
data = """Ana,88
Bo,92
corrupted_line
Cy,79
"""

with open("students.txt", "w") as file:
    file.write(data)
```

```python
process_students("students.txt", "summary.txt")
```

**Explanation**

The function reads each line, splits it using ,, and skips lines that don't contain exactly a name and score. Valid scores are added to total_score, and student_count tracks the number of valid students. The average is then calculated and written to summary.txt.
