**Nested Records to File Report**

**Problem Statement**

*Read employee records from employees.txt, skip malformed records, build a nested dictionary grouped by department, calculate employee count and average salary for each department, and write the results to report.txt.*


**Input**: employees.txt:

Ana,Engineering,7200

Bo,Sales,5400

Cy,Engineering,6800

bad_line_here

Dia,Sales,5900

Eva,Support,4700

**Expected Output : report.txt:**

Engineering: 2 employees, avg salary 7000.00

Sales: 2 employees, avg salary 5650.00

Support: 1 employees, avg salary 4700.00

Console: Skipped malformed record: bad_line_here

**Hints**

Use with open() to read the file.

Use defaultdict(list) to group employees by department.

Each employee should be stored as {"name": name, "salary": salary}.

Use Counter to count departments.

Calculate average salary using sum() / count.

Use writelines() to create the report.

Handle FileNotFoundError.

```python
data = """Ana,Engineering,7200
Bo,Sales,5400
Cy,Engineering,6800
bad_line_here
Dia,Sales,5900
Eva,Support,4700
"""

with open("employees.txt", "w") as file:
    file.write(data)

print("employees.txt created")
```

```python
from collections import defaultdict, Counter

def create_report(input_file, output_file):
    """Read employee records and create a department salary report."""

    employees = defaultdict(list)

    try:
        with open(input_file, "r") as file:
            for line in file:
                line = line.strip()
                parts = line.split(",")

                if len(parts) != 3:
                    print("Skipped malformed record:", line)
                    continue

                name, department, salary = parts

                try:
                    salary = float(salary)
                except ValueError:
                    print("Skipped malformed record:", line)
                    continue

                employees[department].append({
                    "name": name,
                    "salary": salary
                })

    except FileNotFoundError:
        print("Input file not found.")
        return

    # Count employees in each department
    department_counts = Counter()

    for department, employee_list in employees.items():
        department_counts[department] = len(employee_list)

    # Build report
    report = []

    for department, employee_list in employees.items():
        count = department_counts[department]
        average = sum(e["salary"] for e in employee_list) / count

        report.append(
            f"{department}: {count} employees, avg salary {average:.2f}\n"
        )

    with open(output_file, "w") as file:
        file.writelines(report)

    print("Report created successfully.")


create_report("employees.txt", "report.txt")
```
