**DAY-3 : WORKING WITH DICTIONARIES**

**Problem Statement**

*Given a dictionary containing student names and their scores:*

Add a new student.

Update an existing student's score.

Safely look up a student who may not exist.

Print students who scored above 80.


**Input**

scores = {"Ana": 78, "Bo": 85, "Cy": 92}

**Expected Output**

After adding/updating the students:

Bo: 85

Cy: 92

**Hints**

Use scores["name"] = score to add or update.

Use .get() for a safe lookup.

Use .items() to check all students and their scores.

Use > 80 to find students scoring above 80.

```python
scores = {"Ana": 78, "Bo": 85, "Cy": 92}

# Add a new student
scores["Dan"] = 75

# Update an existing student's score
scores["Ana"] = 82

# Safely look up a student
print("Score of Eve:", scores.get("Eve", "Student not found"))

# Print students who scored above 80
print("\nStudents who scored above 80:")

for name, score in scores.items():
    if score > 80:
        print(name, ":", score)
```

**Explanation**

A dictionary stores data as key-value pairs, where the student's name is the key and the score is the value. We use scores["Dan"] = 75 to add a new student and scores["Ana"] = 82 to update an existing score. The .get() method safely checks for a student who does not exist without causing an error, and .items() lets us access every student's name and score.

**Conclusion**

 This exercise demonstrates the basic operations of dictionaries: adding, updating, safely accessing, and filtering data. Dictionaries are useful when we need to store and quickly access related information using a key.
