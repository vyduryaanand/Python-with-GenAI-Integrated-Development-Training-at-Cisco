**PROBLEM STATEMENT**

*Given a nested dict representing a company with departments, each holding a list of employee dicts, print each department's namefollowed by its employees' names, then compute the total employee count across all departments.*

**INPUT**

company = {"Engineering": {"employees": [{"name": "Ana"}, {"name": "Bo"}]},

"Sales": {"employees": [{"name": "Cy"}]}}

**EXPECTED OUTPUT**

Department names with nested employee names printed, plus a total count of 3.

**HINTS**

•Outer dict keys are department names; values are dicts holding a list

•Use a nested for loop: department dict -> employees list -> employee dict

•Accumulate a running total across departments with a counter variable

```python
company = {
    "Engineering": {
        "employees": [{"name": "Ana"}, {"name": "Bo"}]
    },
    "Sales": {
        "employees": [{"name": "Cy"}]
    }
}

total_employees = 0

for department, details in company.items():
    print(department)

    for employee in details["employees"]:
        print(employee["name"])
        total_employees += 1

print("\nTotal employees:", total_employees)
```

**Explanation**

The outer loop goes through each department, while the inner loop goes through the employees list inside that department. employee["name"] prints each employee's name, and total_employees is increased by 1 for every employee.

**Conclusion**

 Nested dictionaries and lists can be accessed using nested loops. This allows us to easily process hierarchical data such as departments and employees.
