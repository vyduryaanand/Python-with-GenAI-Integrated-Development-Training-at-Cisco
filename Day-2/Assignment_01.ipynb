**Log Record Processor: Iterating, Fixed Records & Deduplication**

**Problem Statement**

*Create a Python program that processes raw server log lines, parses each line into a fixed-shape tuple, classifies log levels, tracks unique services, handles extra fields, and produces a sorted summary without modifying the original raw_lines list.*

**Input**

raw_lines = [
    
    "09:01,ERROR,billing,payment failed",
    
    "09:02,INFO,billing,invoice sent",
    
    "09:03,WARNING,auth,slow response",
    
    "09:04,ERROR,auth,token expired,retry=3",
    
    "09:05,INFO,billing,invoice sent",

    ]

**Expected Output**

Line 1: [09:01] ERROR billing -> critical

Line 2: [09:02] INFO billing -> info

Line 3: [09:03] WARNING auth -> warning

Line 4: [09:04] ERROR auth -> critical (extra fields: ['retry=3'])

Line 5: [09:05] INFO billing -> info

Total records: 5

critical: 2, warning: 1, info: 2

Unique services: ['auth', 'billing']

**Hints**

Use split(',') and tuple unpacking with *rest.

Use enumerate() to get line numbers.

Use if/elif to classify each level.

Store unique services in a set.

Use sorted() to create the final service list.

Do not modify raw_lines.

```python
raw_lines = [
    "09:01,ERROR,billing,payment failed",
    "09:02,INFO,billing,invoice sent",
    "09:03,WARNING,auth,slow response",
    "09:04,ERROR,auth,token expired,retry=3",
    "09:05,INFO,billing,invoice sent",
]
```

```python
def process_logs(raw_lines):
    services = set()
    counts = {"critical": 0, "warning": 0, "info": 0}

    for line_no, line in enumerate(raw_lines, start=1):
        timestamp, level, service, message, *rest = line.split(",")

        record = (timestamp, level, service, message)

        if level == "ERROR":
            category = "critical"
        elif level == "WARNING":
            category = "warning"
        else:
            category = "info"

        counts[category] += 1
        services.add(service)

        extra = f" (extra fields: {rest})" if rest else ""

        print(
            f"Line {line_no}: [{record[0]}] "
            f"{record[1]} {record[2]} -> {category}{extra}"
        )

    print(f"Total records: {len(raw_lines)}")
    print(
        f"critical: {counts['critical']}, "
        f"warning: {counts['warning']}, "
        f"info: {counts['info']}"
    )
    print("Unique services:", sorted(services))
```

```python
process_logs(raw_lines)
```

**Explanation**

The program uses split() and extended tuple unpacking to parse each log line into the fixed tuple (timestamp, level, service, message) while storing additional fields in rest. enumerate() provides the line number, if/elif classifies each log level, and a set stores unique service names without duplicates. Finally, sorted() creates the ordered service list without modifying the original raw_lines list.

**Conclusion**

This assignment combines the main Day 2 concepts: tuples and unpacking, extended unpacking, for loops, enumerate(), branching, sets, and sorting. It demonstrates how these built-in Python features can be used together to process structured log data efficiently while preserving the original input.
