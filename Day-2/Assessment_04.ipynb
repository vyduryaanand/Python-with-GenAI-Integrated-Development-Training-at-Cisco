**Problem Statement**

*Write a generator function evens_up_to(n) that yields even numbers from 0 up to n. Use iter() and next() to manually retrieve the first two values, then use a for loop to retrieve the remaining values.*

**Input**

evens_up_to(10)

**Expected Output**

0
2
4
6
8
10

The first two values should be obtained using next(), and the remaining values using a for loop.

**Hints**

A function containing yield automatically becomes a generator.

Use iter() to obtain an iterator.

Use next() to retrieve values one at a time.

Use a for loop to consume the remaining values.

StopIteration occurs when the iterator has no more values.

```python
def evens_up_to(n):
    for number in range(0, n + 1, 2):
        yield number
```

```python
#Create the generator and retrieve the first two values:
numbers = evens_up_to(10)
iterator = iter(numbers)

print(next(iterator))
print(next(iterator))

for number in iterator:
    print(number)
```

**Explanation**

The evens_up_to() function uses yield, which makes it a generator that produces even numbers one at a time. iter() provides the iterator, and next() retrieves the first two values, 0 and 2. The for loop then continues consuming the same iterator and prints the remaining values 4, 6, 8, and 10.

**Conclusion**

This exercise demonstrates Python's iterator protocol and how generators efficiently produce values one at a time. It also shows that manual next() calls and a for loop can be used together to consume the same iterator sequentially.
