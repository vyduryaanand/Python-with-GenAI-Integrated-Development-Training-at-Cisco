**Problem Statement**

*Given a list of GPS ping coordinates represented as (latitude, longitude) tuples, create a dictionary that counts how many times each unique coordinate appears. Then demonstrate what happens when a list is used instead of a tuple as a dictionary key.*

**Input**

pings = [(1.0, 2.0), (1.0, 2.0), (3.0, 4.0)]

**Expected Output**

{(1.0, 2.0): 2, (3.0, 4.0): 1}

When a list is used as a dictionary key:

TypeError: unhashable type: 'list'

**Hints**

Use dict.get(key, 0) + 1 to count occurrences.

Tuples can be used as dictionary keys because they are hashable.

Lists cannot be used as dictionary keys because they are mutable and unhashable.

Try using [lat, lon] as a key to observe the TypeError.

```python
#Counting Coordinate occurences
pings = [
    (1.0, 2.0),
    (1.0, 2.0),
    (3.0, 4.0)
]

counts = {}

for coordinate in pings:
    counts[coordinate] = counts.get(coordinate, 0) + 1

print(counts)
```

```python
#Using a list as a dictionary key
counts = {}

try:
    for lat, lon in pings:
        key = [lat, lon]
        counts[key] = counts.get(key, 0) + 1
except TypeError as e:
    print(e)
```

**Explanation**

The coordinate tuples are used as dictionary keys because tuples are immutable and hashable. The dictionary uses each unique tuple to store its occurrence count. When a list such as [lat, lon] is used as a key, Python raises a TypeError because lists are mutable and therefore unhashable, meaning they cannot be used as dictionary keys.

**Conclusion**

This exercise demonstrates the importance of hashability when choosing dictionary keys. Immutable and hashable objects such as tuples can be used as keys, while mutable objects such as lists cannot. Choosing the appropriate data structure allows GPS coordinates to be efficiently stored and counted.
