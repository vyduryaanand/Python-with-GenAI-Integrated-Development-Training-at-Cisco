**Problem Statement**

*Write a function parse_line(line) that splits a comma-separated string and uses extended unpacking to separate the first two fields from the remaining fields. Return the result as a tuple.*

**Input**

parse_line('Ann,29,Berlin,Germany,10115')

**Expected Output**

name = 'Ann'

age = '29'

rest = ['Berlin', 'Germany', '10115']

**Hints**

Use split(',') to convert the string into a list.

Use *rest to capture all remaining fields.

Return the values as a tuple.

The returned tuple can also be unpacked by the caller.

```python
def parse_line(line):
    fields = line.split(",")
    name, age, *rest = fields
    return name, age, rest
```

```python
#Test the Function
result = parse_line('Ann,29,Berlin,Germany,10115')

name, age, rest = result

print("name =", name)
print("age =", age)
print("rest =", rest)
```

**Explanation**

The split(",") method converts the input string into a list of fields. Extended unpacking assigns the first field to name, the second to age, and uses *rest to collect all remaining fields into a list. The function returns these values as a tuple, which can then be unpacked by the caller.

**Conclusion**

This exercise demonstrates how tuples and extended unpacking can make it easy to separate selected elements from a collection while capturing the remaining elements using *rest. This provides a concise and readable way to process structured data in Python.
