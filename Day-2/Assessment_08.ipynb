**DAY-2: SETS AND THEIR OPERATIONS**

HANDS-ON EXERCISE

**Problem Statement**

*Given two lists of customer IDs from different data sources, use sets to find the IDs present in both sources, IDs only in source A, and IDs present in either source but not both. Deduplicate both lists before performing the operations.*

**Input:**

source_a = ['c1', 'c2', 'c3', 'c2']

source_b = ['c2', 'c3', 'c4']

**Expected Output:**

both = {'c2', 'c3'}

only_a = {'c1'}

symmetric_diff = {'c1', 'c4'}

**Hints**

Convert both lists to sets to remove duplicates.

Use & for intersection.

Use - for difference.

Use ^ for symmetric difference.

Sort the results before printing for consistent output.

```python
source_a = ['c1', 'c2', 'c3', 'c2']
source_b = ['c2', 'c3', 'c4']

set_a = set(source_a)
set_b = set(source_b)

both = set_a & set_b
only_a = set_a - set_b
symmetric_diff = set_a ^ set_b

print("both =", set(sorted(both)))
print("only_a =", set(sorted(only_a)))
print("symmetric_diff =", set(sorted(symmetric_diff)))
```

**Explanation**

The two source lists are first converted into sets, which automatically removes duplicate customer IDs. The intersection operator & finds IDs present in both sources, the difference operator - finds IDs present only in source A, and the symmetric difference operator ^ finds IDs present in either source but not both.

**Conclusion**

This exercise demonstrates how sets and set operations can efficiently handle deduplication and comparison of data from multiple sources. Set operations provide a simple and readable way to identify common, unique, and differing customer IDs.
