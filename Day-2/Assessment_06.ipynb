**Problem Statement**

*Given a list of order dictionaries containing amount and status, write a function that returns a new list containing only paid orders, sorted by amount in descending order, without modifying the original list.*

**Input**

orders = [
    
    {'amount': 50, 'status': 'paid'},
    
    {'amount': 120, 'status': 'pending'},
    
    {'amount': 80, 'status': 'paid'}
]


**Expected Output**

[{'amount': 80, 'status': 'paid'},

 {'amount': 50, 'status': 'paid'}]

The original orders list must remain unchanged.

**Hints**
Use a list comprehension to filter paid orders.

Use sorted() instead of .sort().

Sort using key=lambda o: o['amount'].

Use reverse=True for descending order.

Verify that the original list remains unchanged.

```python
def get_paid_orders(orders):
    paid_orders = [
        order for order in orders
        if order['status'] == 'paid'
    ]

    return sorted(
        paid_orders,
        key=lambda o: o['amount'],
        reverse=True
    )
```

```python
#Test the Function
orders = [
    {'amount': 50, 'status': 'paid'},
    {'amount': 120, 'status': 'pending'},
    {'amount': 80, 'status': 'paid'}
]

original_copy = orders.copy()

result = get_paid_orders(orders)

print(result)
print("Original unchanged:", orders == original_copy)
```

**Explanation**

The function first uses a list comprehension to create a new list containing only orders with the status 'paid'. It then uses sorted() with key=lambda o: o['amount'] and reverse=True to arrange the orders from highest to lowest amount. Since sorted() returns a new list, the original orders list is not modified.

**Conclusion**

This exercise demonstrates safe list manipulation using list comprehensions and sorted(). Creating a new filtered list and sorting it without using .sort() on the original data ensures that the input list remains unchanged while producing the required result.
