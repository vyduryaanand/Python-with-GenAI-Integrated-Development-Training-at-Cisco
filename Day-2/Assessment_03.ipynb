**Problem Statement**

*Write a Python function classify(temp, is_raining) that classifies the weather as 'stay in', 'bring umbrella', or 'go out' using elif and the and/or/in operators.*

**Input**

classify(15, True)

classify(25, False)

classify(5, False)

**Expected Output**

'bring umbrella'

'go out'

'stay in'

**Hints**

Use chained comparisons for temperature ranges.

Combine is_raining with and/or.

Use in where appropriate.

Avoid unnecessary nested if statements.

```python
def classify(temp, is_raining):
    if temp < 10:
        return "stay in"
    elif 10 <= temp <= 20 and is_raining:
        return "bring umbrella"
    elif 10 <= temp <= 20:
        return "go out"
    elif temp > 20 and is_raining:
        return "bring umbrella"
    else:
        return "go out"
```

```python
print(classify(15, True))
print(classify(25, False))
print(classify(5, False))
```

**Explanation**

The function uses if and elif to check temperature conditions and the is_raining value. Chained comparisons such as 10 <= temp <= 20 define the temperature range, while the and operator combines temperature and rain conditions. The function returns 'bring umbrella' when rain is expected in suitable temperatures, 'stay in' when the temperature is below 10°C, and 'go out' for other conditions.

**Conclusion**

This exercise demonstrates how Python's branching and conditional operators can be combined to make decisions based on multiple conditions. Using elif, and, or, and chained comparisons allows the function to remain simple, readable, and easy to understand.
