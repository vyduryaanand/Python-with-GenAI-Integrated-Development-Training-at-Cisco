HANDS-ON EXERCISE : **Refactoring Nested Conditionals and Manual Accumulation**

**Problem Statement**

*Refactor a Python function containing nested if/else conditions and a manual accumulator loop to improve readability while ensuring that its behavior and output remain unchanged.*

**Input**

A list of dictionaries containing user details such as name, age, and active status.

**Expected Output**

A refactored function that filters and transforms the input list and produces identical results to the original function.

**Hints**

Write 2–3 test cases before refactoring.

Ask the AI agent to explain every change.

Consider using a list comprehension or guard clauses.

Do not change the function's public signature.

Compare the outputs before and after refactoring.

```python
#Original Code
def get_active_adults(users):
    result = []

    for user in users:
        if user.get("active"):
            if user.get("age", 0) >= 18:
                if user.get("name"):
                    result.append(user["name"].upper())

    return result
```

```python
#Test Cases
users = [
    {"name": "Vydurya", "age": 22, "active": True},
    {"name": "Rahul", "age": 17, "active": True},
    {"name": "Anil", "age": 25, "active": False},
    {"name": "Priya", "age": 21, "active": True}
]

print(get_active_adults(users))
```

```python
#Refactored Code
def get_active_adults(users):
    return [
        user["name"].upper()
        for user in users
        if user.get("active")
        and user.get("age", 0) >= 18
        and user.get("name")
    ]
```

```python
#Verification
print(get_active_adults(users))
```

**Explanation**

The original function used nested if statements and a manual result accumulator to filter and transform the users. The refactored version replaces this structure with a list comprehension that performs the same checks and converts valid names to uppercase, making the code shorter and easier to read without changing the function signature or behavior.

**Conclusion**

This exercise demonstrates how an AI coding agent can help simplify existing Python code while preserving functionality. By testing the original implementation first and comparing the refactored output, we can verify that the refactoring improves readability without introducing behavioral changes.
