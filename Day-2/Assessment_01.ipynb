**Day-2** :**Refactoring Duplicated Validation Logic**

**Problem Statement**

*Refactor a Python function containing duplicated validation logic to improve readability and maintainability without changing its behavior.*

**Input**
name = "Vydurya"

email = "vydurya@example.com"

age = 22

**Expected Output**

Validation successful

#For invalid input, the function should return:
Validation failed

**Hints**

Identify duplicated validation checks.

Create a reusable helper function.

Use meaningful names.


Do not change the existing
validation rules or output.

Ask the AI agent to refactor the code while preserving behavior.

**AI Prompt**

Refactor this Python function to remove duplicated validation
logic and improve readability without changing its behavior,
validation rules, or output.

```python
#Original Code
def validate_user(name, email, age):
    if not name:
        return False
    if not email:
        return False
    if age < 18:
        return False

    if name == "":
        return False
    if email == "":
        return False
    if age == 0:
        return False

    if len(name) < 2:
        return False
    if "@" not in email:
        return False

    return True
```

```python
#Refactored Code
def validate_user(name, email, age):
    def is_valid(value):
        return value is not None and value != ""

    if not is_valid(name):
        return False

    if not is_valid(email):
        return False

    if age < 18 or age == 0:
        return False

    if len(name) < 2:
        return False

    if "@" not in email:
        return False

    return True
```

```python
#Test Code:
name = "Vydurya"
email = "vydurya@example.com"
age = 22

if validate_user(name, email, age):
    print("Validation successful")
else:
    print("Validation failed")
```

**Explanation**

The original function had repeated checks for empty name and email values. The refactored version uses a reusable is_valid() helper function to perform the common validation, reducing duplication and making the code easier to read while preserving the existing validation behavior.

**Conclusion**

This activity demonstrates how GenAI can assist in refactoring existing Python code. By identifying duplicated logic and suggesting a reusable helper function, the AI agent helps improve code quality, readability, and maintainability without changing the intended functionality.
