**DAY 1 · CONVERSATIONAL CODING & PROMPT ENGINEERING FOR PYTHON**

**Problem Statement**

Create three prompts of increasing quality—Weak, Better, and Expert—for the same task: writing a Python function that reads a log file and counts entries based on severity: INFO, WARNING, and ERROR.

**Input**

Parse a log file and count entries by severity level: INFO, WARNING, ERROR.

**Expected Output**

Three prompts: Weak → Better → Expert
Python code generated using the Expert prompt
Brief review of the generated code.

**Hints**

The Expert prompt should specify the function name and parameters.

Mention the expected log format.

Specify what the function should return.

Ask for a docstring.

Ask the AI to handle malformed lines.

**Code**

**Prompt 1 — Weak**

*Write a Python function to parse a log file and count errors by severity.*

**Prompt 2 — Better**

*Write a Python function that reads a log file and counts the number of INFO, WARNING, and ERROR entries. Return the counts for each severity level*.

**Prompt 3 — Expert**

*Write a Python function count_log_severity(file_path) that reads a text log file where each line contains a timestamp, severity level (INFO, WARNING, or ERROR), and message, separated by spaces. Return a dictionary with counts for INFO, WARNING, and ERROR. Include a docstring, ignore malformed lines, and handle the case where the file does not exist. Keep the solution simple and readable.*

```python
log_data = """10:00:01 INFO Server started
10:01:15 ERROR Database failed
10:02:30 WARNING High memory usage
10:03:10 ERROR Connection failed
10:04:00 INFO Request completed
"""

with open("sample.log", "w") as file:
    file.write(log_data)

print("sample.log created successfully!")
```

```python
def count_log_severity(file_path):
    counts = {
        "INFO": 0,
        "WARNING": 0,
        "ERROR": 0
    }

    try:
        with open(file_path, "r") as file:
            for line in file:
                parts = line.strip().split()

                if len(parts) < 3:
                    continue

                severity = parts[1]

                if severity in counts:
                    counts[severity] += 1

    except FileNotFoundError:
        print("File not found.")

    return counts


result = count_log_severity("sample.log")

print("Log Severity Counts:", result)
```

Explanation

**The three prompts become more specific step by step. The Weak prompt gives only the basic task, so the AI has to make many assumptions. The Better prompt provides the required severity levels and expected result. The Expert prompt gives the function name, input format, return structure, documentation requirement, and error handling, allowing the AI assistant to generate a more precise solution.**

Conclusion

**Better prompts generally produce code that requires less manual correction because they clearly describe the expected behavior, inputs, outputs, and edge cases. This exercise demonstrates that specific and well-structured prompts lead to more useful AI-generated Python code.**
