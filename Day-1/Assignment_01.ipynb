**AI-Assisted Log Line Processor**

**Problem Statement**

*Create a Python program that processes application log lines. The program should clean and split each log line into date, level, and message, count how many lines contain error, demonstrate shallow and deep copying, and encode/decode the final summary using UTF-8.*

**SAMPLE INPUT**

raw_lines = [

" 2024-06-10 ERROR disk quota exceeded ",

"2024-06-10 INFO backup completed",

" 2024-06-10 error connection refused",

]

**EXPECTED OUTPUT**

parsed: [{'date': '2024-06-10', 'level': 'ERROR', 'message': 'disk quota exceeded'}, ...]

error_count: 2

shallow copy shares nested objects with original (proven via id())

deep copy is fully independent (proven via id())

3 passed in pytest
b'...' (encoded bytes) and matching decoded string

```python
#Import and Create the Input
import copy

raw_lines = [
    " 2024-06-10 ERROR disk quota exceeded ",
    "2024-06-10 INFO backup completed",
    " 2024-06-10 error connection refused",
]

print(raw_lines)
```

```python
#Create parse_log_line()
def parse_log_line(raw: str) -> dict:
    """Parse a log line into date, level, and message."""

    parts = raw.strip().split()

    if len(parts) < 2:
        return {
            "date": "",
            "level": "",
            "message": ""
        }

    return {
        "date": parts[0],
        "level": parts[1],
        "message": " ".join(parts[2:])
    }
```

```python
#Parse Logs and Count Errors
parsed = []

for line in raw_lines:
    parsed.append(parse_log_line(line))

error_count = 0

for line in raw_lines:
    if "error" in line.lower():
        error_count += 1

print("parsed:", parsed)
print("error_count:", error_count)
```

```python
#Demonstrate Shallow and Deep Copy

shallow = copy.copy(parsed)
deep = copy.deepcopy(parsed)

print("Original ID:", id(parsed))
print("Shallow ID:", id(shallow))
print("Deep ID:", id(deep))

print("\nNested object IDs:")
print("Original first dict:", id(parsed[0]))
print("Shallow first dict:", id(shallow[0]))
print("Deep first dict:", id(deep[0]))

# Mutate the shallow copy
shallow[0]["level"] = "WARNING"

print("\nAfter changing shallow copy:")
print("Original:", parsed)
print("Shallow:", shallow)
print("Deep:", deep)
```

```python
#UTF-8 Encoding and Decoding
summary = f"Processed {len(parsed)} log lines, found {error_count} errors."

encoded = summary.encode("utf-8")
decoded = encoded.decode("utf-8")

print(encoded)
print(decoded)
```
