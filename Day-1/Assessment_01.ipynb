**PROBLEM STATEMENT**

*Write a script that demonstrates aliasing vs copying: create a list, alias it, shallow-copy it, and deep-copy a nested version, printing id() and contents at each step to prove which operations share memory.*

**INPUT**

A starting list such as data = [1, 2, [3, 4]]

**EXPECTED OUTPUT**

Printed id() values showing alias shares id with original; shallow copy has a different outer id but shares the nested list's id; deep copy shares neither.

**HINTS** (no full solution here —give it a try first!)

*   Use id() before and after each copy operation
*   Mutate the nested list through each copy and observe which other variables  see the change
*  Use copy.deepcopy for the fully independent version










```python
import copy

# Original list
data = [1, 2, [3, 4]]

# Aliasing
alias = data

# Shallow copy
shallow = data.copy()

# Deep copy
deep = copy.deepcopy(data)

# Display contents
print("Original:", data)
print("Alias:", alias)
print("Shallow Copy:", shallow)
print("Deep Copy:", deep)

print("\n--- IDs ---")

print("Original ID:", id(data))
print("Alias ID:", id(alias))
print("Shallow Copy ID:", id(shallow))
print("Deep Copy ID:", id(deep))

print("\n--- Nested List IDs ---")

print("Original Nested ID:", id(data[2]))
print("Alias Nested ID:", id(alias[2]))
print("Shallow Nested ID:", id(shallow[2]))
print("Deep Nested ID:", id(deep[2]))
```

**Explanation**

Aliasing means both variables point to the same object, so data and alias have the same ID. In a shallow copy, a new outer list is created, so its ID is different, but the nested list is still shared. In a deep copy, both the outer list and nested list are copied, so their IDs are different.

**Conclusion**

Aliasing shares everything, shallow copying creates a new outer list but shares nested objects, and deep copying creates a completely independent copy.
