*For three hypothetical projects (a real-time* *trading engine, an internal automation script, and* *an LLM-powered chatbot), decide whether* *Python is the right primary language and justify* *each decision in one sentence.*

**INPUT**

Project descriptions:
(1) sub-millisecond trading engine,
(2) nightly log-cleanup automation,
(3) customer-support chatbot using an LLM API.

**EXPECTED OUTPUT**

A short written justification per project (e.g. 'No for #1 due to GIL/latency; Yes for #2 and #3 due to iteration speed and AI SDK support').

**HINTS (no full solution here —give it a try first!)**

```python
projects = [
    "Sub-millisecond trading engine",
    "Nightly log-cleanup automation",
    "Customer-support chatbot using an LLM API"
]

decisions = [
    "No - Extremely low latency and high performance are required.",
    "Yes - Python is well suited for automation and scripting.",
    "Yes - Python has strong support for LLM APIs and AI libraries."
]

for project, decision in zip(projects, decisions):
    print(project)
    print(decision)
    print()
```

**Explanation**

Python is not the ideal primary language for a sub-millisecond trading engine because such systems require extremely low latency and predictable high performance. However, Python is a good choice for nightly automation because it is simple and has excellent scripting support, and it is also suitable for an LLM-powered chatbot because of its strong AI ecosystem and support for LLM APIs.

**Conclusion**

The choice of programming language depends on the project's requirements. Python is generally strong for automation, APIs, AI, and rapid development, while highly performance-critical systems may require languages designed for lower latency and greater execution performance.
