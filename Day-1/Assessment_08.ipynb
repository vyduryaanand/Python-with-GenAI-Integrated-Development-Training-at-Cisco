**PROBLEM STATEMENT**

Given a raw log line with extra whitespace and mixed case, write a script that strips it, checks for the substring 'error' case-insensitively, splits it into tokens, rejoins them with a pipe delimiter, and prints both the UTF-8 encoded bytes and the decoded string.

**INPUT**

" 2024-06-10 Error disk quota EXCEEDED "

**EXPECTED OUTPUT**

2024-06-10|Error|disk|quota|EXCEEDED

b'2024-06-10|Error|disk|quota|EXCEEDED'

true (matches decoded)

**HINTS(no full solution here —give it a try first!)**

•Use .strip(), .lower(), and 'in' for the case-insensitive check

•Use .split() and '|'.join() to rebuild the tokenized string

•Use .encode('utf-8') and .decode('utf-8') to round-trip the string

```python
# Raw log line
log_line = " 2024-06-10 Error disk quota EXCEEDED "

# Remove extra spaces
cleaned_line = log_line.strip()

# Check for 'error' ignoring case
contains_error = "error" in cleaned_line.lower()

# Split into tokens
tokens = cleaned_line.split()

# Join tokens using pipe delimiter
joined_string = "|".join(tokens)

# Encode into UTF-8 bytes
encoded_bytes = joined_string.encode("utf-8")

# Decode back into string
decoded_string = encoded_bytes.decode("utf-8")

# Print results
print(joined_string)
print(encoded_bytes)
print(decoded_string)

# Check if decoded string matches original joined string
print(joined_string == decoded_string)
```

**Explanation**

The program first removes the extra spaces using .strip(). It then converts the log line to lowercase and checks whether error is present. Using .split() and '|'.join(), it converts the log into a pipe-separated string. Finally, .encode("utf-8") converts the string into bytes, and .decode("utf-8") converts those bytes back into a string.
