```python
def remove_duplicates(items):
    seen = set()
    result = []

    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)

    return result


numbers = [1, 2, 2, 3, 1, 4]

print("Original:", numbers)
print("After removing duplicates:", remove_duplicates(numbers))
```

**Explanation**

The function uses a set called seen to remember which values have already appeared. When a new value is found, it is added to seen and result. If the value has already been seen, it is skipped. This removes duplicates while keeping the original order.
