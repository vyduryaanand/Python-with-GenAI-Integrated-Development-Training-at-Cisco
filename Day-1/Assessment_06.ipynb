```python
%%writefile test_discount.py

def calculate_discount(price: float, discount: float) -> float:
    """
    Calculate the price after applying a percentage discount.
    """
    return price - (price * discount / 100)


def test_normal_discount():
    assert calculate_discount(100, 20) == 80


def test_zero_discount():
    assert calculate_discount(100, 0) == 100


def test_negative_price():
    assert calculate_discount(-100, 20) == -80
```

```python
!pytest -v test_discount.py
```

**Explanation**

The AI assistant was used to improve the original function by adding type hints and a docstring, and to generate multiple tests covering normal and edge-case behavior. The tests were then placed into a pytest file and executed in Google Colab using pytest -v.
