*Write a script that prompts the user for their name and age, converts age to int, and prints a formatted greeting on one lineusing print()'s sep/end parameters. Then use type() and dir() to explore the type of the age variable.*

**INPUT**

Interactive input, e.g. name='Sam', age='30'

**EXPECTED OUTPUT**

Hello, Sam -you are 30 years old!

<class 'int'>

**HINTS**(no full solution here —give it a try first!)

•Remember input() returns a str —convert age explicitly

•Use sep and end to control formatting without string concatenation

•Call dir(age) and filter out dunder attributes for a cleaner list


```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("Hello,", name, "- you are", age, "years old!", sep=" ", end="\n")

print(type(age))

print([item for item in dir(age) if not item.startswith("__")])
```

**Explanation**

input() gives the age as a string, so int() converts it into an integer. The print() statement uses sep=" " to put spaces between the values and end="\n" to move to the next line. type(age) confirms that the age is an integer, while dir(age) shows the methods and attributes available for an integer; the list comprehension removes the special __...__ items.
